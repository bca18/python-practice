Built in functions

```python
# Given a list of nums, this is unsorted list
# Requirement - I need second largest element in the list

nums = [23, 45, 12, 78, 36, 100, 1000] # Second largest = 45
```

```python
# Sort it first = [12, 23, 36, 45, 78, 100, 1000]
# sorted
sorted_nums = sorted(nums)
sorted_nums[-2]
```

```python
len(nums)-1
```

```python
sorted_nums[len(nums)-2]
```

Problem - Given you have unsorted list of nums, We need second largest element
Algorithm:
1) Sort the list
2) Use indexing - Reverse indexing (-2) to get second largest element

Alternative:
3) Use len() function - to get the length of the list, len(nums)-2 gives me second largest element.
